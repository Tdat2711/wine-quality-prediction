# Xử lý missing values — Dương

Notebook này chỉ xử lý 9 cột có missing values. Không xóa duplicate, xử lý outlier, chuẩn hóa văn bản hay đổi kiểu dữ liệu.

```python
from pathlib import Path
import sys
import pandas as pd

project_root = Path.cwd().resolve()
if project_root.name == 'notebooks':
    project_root = project_root.parent
sys.path.insert(0, str(project_root))

from src.handle_missing_values import MISSING_COLUMNS, handle_missing_values

input_path = project_root / 'data' / 'raw' / 'winemag-data-130k-v2.csv'
output_path = project_root / 'data' / 'processed' / 'wine_missing_values_handled.csv'
df = pd.read_csv(input_path)
df.shape
```

```python
missing_before = (
    df.loc[:, MISSING_COLUMNS]
      .isna()
      .sum()
      .sort_values(ascending=False)
)
missing_before
```

```python
cleaned, report = handle_missing_values(df)
pd.Series(report, name='Số ô xử lý')
```

```python
missing_after = cleaned.loc[:, MISSING_COLUMNS].isna().sum()
unchanged_columns = [column for column in df.columns if column not in MISSING_COLUMNS]

assert cleaned.shape == df.shape
assert cleaned.columns.tolist() == df.columns.tolist()
assert int(missing_after.sum()) == 0
assert cleaned[unchanged_columns].equals(df[unchanged_columns])

pd.DataFrame({'Trước': missing_before, 'Sau': missing_after})
```

```python
output_path.parent.mkdir(parents=True, exist_ok=True)
cleaned.to_csv(output_path, index=False)
output_path
```
